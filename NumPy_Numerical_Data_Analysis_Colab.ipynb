# NumPy for Numerical Data Analysis

This notebook demonstrates the core NumPy concepts used for numerical data analysis:

- Creating 1D, 2D, and 3D arrays
- Checking dimensions, shape, and data types
- Indexing and slicing
- Reshaping arrays
- Generating arrays with `zeros()`, `ones()`, `arange()`, and `linspace()`
- Mathematical and vectorized operations
- Boolean masking
- Broadcasting
- Aggregate functions such as `sum()`, `mean()`, `min()`, `max()`, and `median()`

**Goal:** Understand how NumPy makes numerical computations efficient and concise.

## 1. Import NumPy

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Creating 1D, 2D, and 3D Arrays

```python

array_1d = np.array([10, 20, 30, 40, 50])

array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:")
print(array_1d)

print("\n2D array:")
print(array_2d)

print("\n3D array:")
print(array_3d)
```

## 3. Checking Dimensions, Shape, Size, and Data Type

```python
print("1D ndim:", array_1d.ndim)
print("1D shape:", array_1d.shape)
print("1D size:", array_1d.size)
print("1D dtype:", array_1d.dtype)

print("\n2D ndim:", array_2d.ndim)
print("2D shape:", array_2d.shape)
print("2D size:", array_2d.size)
print("2D dtype:", array_2d.dtype)

print("\n3D ndim:", array_3d.ndim)
print("3D shape:", array_3d.shape)
print("3D size:", array_3d.size)
print("3D dtype:", array_3d.dtype)
```

## 4. Indexing and Slicing

```python

print("First element:", array_1d[0])
print("Last element:", array_1d[-1])

print("Elements from index 1 to 3:", array_1d[1:4])
print("Every second element:", array_1d[::2])

print("\n2D element at row 1, column 2:", array_2d[1, 2])

print("First row:", array_2d[0, :])
print("Second column:", array_2d[:, 1])
print("First two columns:")
print(array_2d[:, :2])

print("\n3D element [1, 0, 1]:", array_3d[1, 0, 1])
```

## 5. Reshaping Arrays

```python
original = np.arange(1, 13)

print("Original array:")
print(original)

reshaped = original.reshape(3, 4)

print("\nReshaped to 3 x 4:")
print(reshaped)

reshaped_3d = original.reshape(2, 2, 3)

print("\nReshaped to 2 x 2 x 3:")
print(reshaped_3d)
```

## 6. Generating Arrays with NumPy Functions

```python

zeros_array = np.zeros((2, 3))


ones_array = np.ones((2, 3))


range_array = np.arange(0, 10, 2)

linear_array = np.linspace(0, 1, 5)

print("zeros():")
print(zeros_array)

print("\nones():")
print(ones_array)

print("\narange(0, 10, 2):")
print(range_array)

print("\nlinspace(0, 1, 5):")
print(linear_array)
```

## 7. Mathematical Operations

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)

print("\nSquare of a:", a ** 2)
print("Square root of a:", np.sqrt(a))
print("Natural logarithm of a:", np.log(a))
```

## 8. Vectorized Operations

```python
.

temperatures = np.array([20, 22, 25, 28, 30])

celsius_to_fahrenheit = temperatures * 9/5 + 32

print("Temperatures in Celsius:", temperatures)
print("Temperatures in Fahrenheit:", celsius_to_fahrenheit)

print("\nTemperatures above 24°C:", temperatures > 24)
```

## 9. Boolean Masking

```python
scores = np.array([45, 67, 82, 91, 56, 74, 38, 88])


passed = scores >= 50

print("Scores:", scores)
print("Boolean mask (score >= 50):", passed)

print("Passing scores:", scores[passed])


print("Scores above 75:", scores[scores > 75])


print("Scores between 60 and 90:",
      scores[(scores >= 60) & (scores <= 90)])
```

## 10. Broadcasting

```python

data = np.array([10, 20, 30, 40])

print("Original data:", data)
print("Add 5:", data + 5)
print("Multiply by 2:", data * 2)


matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

offset = np.array([1, 2, 3])

print("\nMatrix:")
print(matrix)

print("\nOffset:")
print(offset)

print("\nMatrix + offset (broadcasting):")
print(matrix + offset)
```

## 11. Aggregate Functions

```python
data = np.array([10, 20, 30, 40, 50, 60])

print("Data:", data)
print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Median:", np.median(data))
```

## 12. Aggregate Functions Along an Axis

```python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Matrix:")
print(matrix)

print("\nSum of all elements:", np.sum(matrix))
print("Column-wise sum (axis=0):", np.sum(matrix, axis=0))
print("Row-wise sum (axis=1):", np.sum(matrix, axis=1))

print("\nColumn-wise mean:", np.mean(matrix, axis=0))
print("Row-wise mean:", np.mean(matrix, axis=1))

print("\nColumn-wise minimum:", np.min(matrix, axis=0))
print("Row-wise maximum:", np.max(matrix, axis=1))

print("\nColumn-wise median:", np.median(matrix, axis=0))
```

## 13. Mini Numerical Data Analysis Example

Let's use NumPy to analyze the marks of five students in three subjects.

```python
marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [92, 88, 95],
    [55, 60, 68],
    [81, 79, 84]
])

print("Marks (Rows = students, Columns = subjects):")
print(marks)

print("\nAverage mark of each student:")
print(np.mean(marks, axis=1))

print("\nAverage mark in each subject:")
print(np.mean(marks, axis=0))

print("\nHighest mark in each subject:")
print(np.max(marks, axis=0))

print("\nLowest mark in each subject:")
print(np.min(marks, axis=0))

student_averages = np.mean(marks, axis=1)
print("\nStudents with average >= 75:")
print(student_averages[student_averages >= 75])
```
